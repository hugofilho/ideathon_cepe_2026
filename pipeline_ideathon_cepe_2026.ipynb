```python
df = pd.read_excel(excel_path, sheet_name='Avaliaacao_Ideathon_ Out_26')
```

```python
print("Perfis encontrados:")
print(df['Qual é o seu perfil de atuação no evento?'].value_counts())

print("\n--- Vínculo por perfil ---")
print(df.groupby('Qual é o seu perfil de atuação no evento?')['Qual é o seu vínculo principal no evento?'].value_counts(dropna=False))
```

```python
# Estatísticas dos estudantes/comunidade (34 respostas)
participantes = df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Participante', na=False)]
print("\n--- Participantes (Estudante / Comunidade) ---")
cols_part = [c for c in df.columns if df[c].notna().sum() >= 30 and c not in ['Carimbo de data/hora', 'Qual é o seu perfil de atuação no evento?', 'Qual é o seu vínculo principal no evento?']]
for col in cols_part:
    if df[col].dtype == 'float64':
        print(f"{col}: Média = {participantes[col].mean():.2f}, DesvPad = {participantes[col].std():.2f}, Min = {participantes[col].min()}, Max = {participantes[col].max()}")
    else:
        print(f"\n{col}:")
        print(participantes[col].value_counts())

# Moderadores
mod = df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Moderador', na=False)]
print(f"\n--- Moderadores (N = {len(mod)}) ---")
for col in df.columns:
    if mod[col].notna().sum() > 0:
        print(f"{col}: {mod[col].tolist()}")

# Profissionais / Gestores / Demandantes
prof = df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Profissional|Demandante|Gestor', na=False)]
print(f"\n--- Profissionais / Gestores (N = {len(prof)}) ---")
for col in df.columns:
    if prof[col].notna().sum() > 0:
        print(f"{col}: {prof[col].tolist()}")
```

```python
# Vamos analisar detalhadamente a distribuição dos dados

print("=== DISTRIBUIÇÃO COMPLETA DE PERFIS E VÍNCULOS ===")
print(df[['Qual é o seu perfil de atuação no evento?', 'Qual é o seu vínculo principal no evento?']].value_counts(dropna=False))

print("\n=== PARTICIPANTES / ESTUDANTES / MORADORES / PROFISSIONAIS (N=34) ===")
# Detalhamento por vínculo dentro do grupo participante
print("Por vínculo:")
print(df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Participante')]['Qual é o seu vínculo principal no evento?'].value_counts())

print("\n--- Médias Gerais dos Participantes (Escala Likert 1-5) ---")
p_cols = [
    'Como você avalia a experiência de propor soluções voltadas para a saúde de Ceres (GO)? [Aprendizado sobre os problemas da saúde local]',
    'Como você avalia a experiência de propor soluções voltadas para a saúde de Ceres (GO)? [Integração com a equipe]',
    'Como você avalia a experiência de propor soluções voltadas para a saúde de Ceres (GO)? [Desafio de criar uma solução em curto tempo]',
    'Qual a probabilidade de você continuar desenvolvendo a solução após o evento?',
    'O Ideathon aumentou seu interesse em participar de projetos de inovação social/saúde na região de Ceres?'
]
for col in p_cols:
    short_name = col.split('[')[-1].replace(']', '') if '[' in col else col.split('?')[-1]
    val = df[col].dropna()
    print(f"- {short_name}: Média = {val.mean():.2f} (±{val.std():.2f}), Mediana = {val.median()}")

print("\n--- Diferencial das Soluções (Visão Participante) ---")
print(df['Na sua visão como participante/comunidade, qual é o maior diferencial da solução desenvolvida pela sua equipe?'].value_counts())

print("\n=== DEMANDANTES / GESTORES / PROFISSIONAIS (N=2) ===")
d_cols = [
    'As soluções apresentadas atenderam às demandas reais e prioritárias da rede de saúde local?',
    'Avalie o potencial de adoção das propostas pela gestão pública ou instituições locais: [Viabilidade financeira de implementação]',
    'Avalie o potencial de adoção das propostas pela gestão pública ou instituições locais: [Alinhamento com as normas legais do SUS]',
    'Avalie o potencial de adoção das propostas pela gestão pública ou instituições locais: [Facilidade de adesão pelas equipes de saúde atuais]'
]
demandantes = df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Demandante')]
for col in d_cols:
    short_name = col.split('[')[-1].replace(']', '') if '[' in col else col.split('?')[-1]
    vals = demandantes[col].tolist()
    print(f"- {short_name}: {vals} (Média: {pd.Series(vals).mean():.2f})")

print("\n=== MODERADORES / MENTORES (N=2) ===")
m_cols = [
    'Em relação ao desempenho das equipes que você acompanhou, como você avalia os pontos abaixo? [Compreensão dos gargalos da saúde em Ceres]',
    'Em relação ao desempenho das equipes que você acompanhou, como você avalia os pontos abaixo? [Capacidade de resposta aos feedbacks]',
    'Em relação ao desempenho das equipes que você acompanhou, como você avalia os pontos abaixo? [Evolução do projeto do início ao fim]',
    'As soluções propostas apresentaram fundamentação coerente com a realidade do SUS em Ceres?',
    'Qual é o nível de maturidade médio dos projetos apresentados para avançarem para uma fase de incubação/piloto?'
]
moderadores = df[df['Qual é o seu perfil de atuação no evento?'].str.contains('Moderador')]
for col in m_cols:
    short_name = col.split('[')[-1].replace(']', '') if '[' in col else col.split('?')[-1]
    vals = moderadores[col].tolist()
    print(f"- {short_name}: {vals} (Média: {pd.Series(vals).mean():.2f})")
```
